# 1. Bibliotheken importieren und Daten laden 💾


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.manifold import TSNE
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn import set_config
set_config(transform_output='pandas')
```

```python
songs_df = pd.DataFrame()
url = 'https://drive.google.com/file/d/1l9eV0mK87ACovb9ySvvyAwHnFwtmUlZx/view?usp=sharing'
path = 'https://drive.google.com/uc?export=download&id='+url.split('/')[-2]
songs_df = pd.read_csv(path)
songs_df
```

# Leerzeichen am Anfang und Ende der Spaltennamen entfernen

```python
songs_df.columns.to_list()
```

```python
songs_df.columns = songs_df.columns.str.strip()
songs_df.set_index('name', inplace=True)
```

# Die Spalten `Unnamed`, `id`, `type`, `artist` und `html` entfernen, da sie keine numerischen Merkmale für das Clustering enthalten

```python
columus_drop = ['Unnamed: 0','artist','type','id', 'html']
songs_copy_df = songs_df.drop(columns= columus_drop).copy()
```

```python
songs_copy_df.info()
```

```python
songs_copy_df
```

# Daten mit StandardScaler skalieren
  Mit StandardScaler skalieren wir die musikalischen Merkmale, sodass sie einen Mittelwert von ungefähr 0 und eine Standardabweichung von 1 haben. Dadurch verhindern wir, dass Merkmale mit größeren Zahlenwerten die Distanzberechnung bei DBSCAN übermäßig beeinflussen.

```python
# StandardScaler initialisieren
my_standardScaler = StandardScaler()

# Musikalische Merkmale standardisieren
scaled_songs_df = my_standardScaler.fit_transform(songs_copy_df)

# Skalierte Daten anzeigen
scaled_songs_df
```

# Agglomerative Clustering 🔍

## Die Anzahl der Cluster auswählen

```python
linked = linkage (scaled_songs_df, method='ward')

plt.figure(figsize=(24, 12))
dendrogram(
    linked,
    orientation= 'top',
    distance_sort= True,
    show_leaf_counts= False
)
plt.title('Dendrogram')
plt.xlabel('Data Points')
plt.xticks(rotation=90)
plt.ylabel('Euclidean Distance')
plt.show()
```

Die Anzahl der Cluster kann man mithilfe eines horizontalen Schnitts im Dendrogramm bestimmen. Die Anzahl der Schnittpunkte mit den vertikalen Ästen entspricht der Anzahl der Cluster.

# Agglomerative Clustering anwenden

```python
clusters_number = 9

agglo_clustering = AgglomerativeClustering(
                                          n_clusters= clusters_number,
                                          metric= 'euclidean',
                                          linkage='ward'
                                        )
agglo_clustering.fit(scaled_songs_df)
scaled_songs_df['Cluster'] = agglo_clustering.labels_
```

```python
songs_number_in_clusters = scaled_songs_df['Cluster'].value_counts().sort_index()
songs_number_in_clusters
```

# Visualise Clusters using t-SNE 📊

```python
# Initialise t-SNE with 2 components
tsne = TSNE(n_components=2,
            random_state=42,)

# Fit and transform the scaled data (excluding the Cluster column)
tsne_components = tsne.fit_transform(scaled_songs_df.drop('Cluster', axis=1))

# Add the Cluster labels to the t-SNE DataFrame
tsne_components['Cluster'] = scaled_songs_df['Cluster']
tsne_components['Cluster'] = tsne_components['Cluster'].astype(str)

(
sns.relplot(data = tsne_components,
            x = 'tsne0',
            y = 'tsne1',
            hue = 'Cluster',
            palette = 'viridis')
.set(title = "Hierarchical Clustering Visualisation with t-SNE")
);
```

Das Agglomerative Clustering hat 9 Cluster erstellt. Die Anzahl der Songs ist jedoch unterschiedlich verteilt. Cluster 4 ist mit 1.435 Songs am größten, während Cluster 8 mit 52 Songs am kleinsten ist.

Die t-SNE-Visualisierung zeigt die räumliche Verteilung der Songs in zwei Dimensionen.
